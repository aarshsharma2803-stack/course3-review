# 7.2 Privacy and Ethics for Higher Ed Data — Code Brief

## Key Concepts

- **Administrative data isn't "free data"** — collected for program administration, not research; secondary use for ML introduces ethical tension (44 USC §3561 statistical vs. non-statistical purposes).
- **ASA Ethical Guidelines** (8 domains) + **ACM Code of Ethics** govern ML practice. Focus domains: Integrity of Data and Methods (B), Responsibilities to Stakeholders (C), Responsibilities to Data Subjects (D).
- **Two dimensions of ethical practice**:
  1. **Practice ethically** — technical integrity in daily work (transparency, fairness, documentation)
  2. **Identify and respond to unethical requests** — professional courage
- **Stakeholder Analysis Framework** — map harm/benefit/unknowns across analyst, supervisor, leadership, students, faculty, public trust. Harms are not equal or exchangeable.
- **6-Step Ethical Reasoning**: (1) Identify prerequisite knowledge (2) Identify decision framework (3) Recognize the ethical issue (4) Evaluate alternatives (5) Make & justify decision (6) Reflect & document.
- **Governance mechanisms**: Model Cards, Institutional Review Processes, Data Governance Boards, Reflection Cycles.

No code in this lecture — pure conceptual/ethics framework.
