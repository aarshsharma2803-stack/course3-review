# 7.2 Privacy and Ethics for Higher Ed Data
- Administrative data isn't "free data"
- ASA / ACM ethical frameworks
- Stakeholder analysis
- 6-step ethical reasoning

## Administrative Data Isn't "Free Data"
Data collected to administer a program (retention tracking, LMS logs, financial aid) wasn't collected for ML. Using it for predictive risk scoring is a *secondary use* — that introduces real ethical tension, not just a technical step.

## Governing Frameworks
- **ASA Ethical Guidelines** (8 domains) — focus on Integrity of Methods, Responsibilities to Stakeholders, Responsibilities to Data Subjects
- **ACM Code of Ethics** — public good, avoid harm, respect privacy, transparency

Both apply directly to your ML pipeline, from data split to deployment.

## Two Dimensions of Ethical Practice
1. **Practice ethically** — technical integrity in daily work (transparent missing-data handling, fair model choices)
2. **Identify and respond to unethical requests** — professional courage (e.g. being asked to "make the numbers look better")

## Stakeholder Analysis
Map harm/benefit/unknowns across: you, your supervisor, leadership, students, faculty, public trust. Two rules: **all harms are not equal**, and **harms and benefits are not exchangeable** — you can't offset one group's harm with another's benefit.

## 6-Step Ethical Reasoning
1. Identify prerequisite knowledge (ASA/ACM, FERPA/HEOA, stakeholder analysis)
2. Identify the decision framework
3. Recognize the ethical issue
4. Evaluate alternatives
5. Make & justify the decision
6. Reflect and document
