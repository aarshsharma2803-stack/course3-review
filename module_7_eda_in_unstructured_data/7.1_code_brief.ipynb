# 7.1 Why Surveys in Higher Ed — Code Brief

## Key Concepts

- Surveys capture **student voice**: engagement, belonging, satisfaction, perceived learning — constructs administrative data (GPA, enrollment) can't capture.
- **National surveys**: NSSE, SERU, CIRP, CCSSE — validated instruments, benchmarkable across institutions.
- **Internal surveys**: first-year experience, advising satisfaction, exit surveys, stop-out surveys — fast to deploy, need consistent measurement.
- **Survey data structure**: rows = respondents (Sampling Units), columns = item responses (Features) — Likert/numeric (ordinal), categorical (needs one-hot encoding), free text (needs vectorization).
- **UPAD cycle applied to surveys**:
  - **Understand** — define constructs, response scales
  - **Prepare** — clean missing responses, encode Likert scales, handle non-response bias, transform text
  - **Analyze** — explore distributions, assess correlations, build predictive models
  - **Deploy** — inform policy, target interventions, monitor trends

No code in this lecture — pure conceptual foundation. 7.3 introduces the first vectorization code.
