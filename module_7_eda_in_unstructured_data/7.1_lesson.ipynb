# 7.1 Why Surveys in Higher Ed
- Student voice beyond administrative data
- National vs. internal surveys
- Rows = respondents, columns = items
- UPAD cycle applied to survey data

## Why Surveys Matter
Administrative data tells you *what* happened (GPA, retention). Surveys tell you *how it felt and why* — engagement, belonging, stress, perceived support. Two students with identical GPA/retention can have very different survey answers, and that difference matters for intervention.

## National vs. Internal Surveys
- **National**: NSSE, SERU, CIRP, CCSSE — validated, benchmarkable across institutions
- **Internal**: first-year experience, advising satisfaction, exit surveys — fast to build, need consistent measurement

## Survey Data Structure
- **Rows** = respondents (one student per row, joins to admin data)
- **Columns** = item responses — Likert/numeric (ordinal, usable directly), categorical (needs one-hot), free text (needs vectorization — that's 7.3)

## UPAD Cycle for Surveys
- **Understand** — define constructs, response scales
- **Prepare** — clean missing, encode Likert, handle non-response bias, vectorize text
- **Analyze** — distributions, correlations, models
- **Deploy** — inform policy, target interventions, monitor trends
