# 8.1 Introduction to Machine Learning with Survey Data — Code Brief

## Key Concepts

- Survey data capture the "Human Element" administrative records miss: engagement, belonging, perceived support, stress — early indicators of future outcomes.
- **Supervised learning in a survey context**: target ($y$) = academic success/failure, attitude categories, performance scores, persistence. Features ($X$) = Likert responses, binary items, one-hot categoricals, composite scale scores, text vectors (from Module 7).
- **Common algorithm families**: Linear models (Logistic Regression — most interpretable), Tree-based (Decision Trees — intuitive rules), Ensembles (Random Forest/XGBoost — highest predictive power), Kernel methods (SVM), Neural Networks (MLP — highest capacity, least interpretable).
- No single "best" algorithm — the winner depends on survey instrument, population size, outcome type, and missing-data patterns.
- **Literature review takeaway**: across 3 real studies (academic performance, early-warning systems, student stress), ensemble methods (especially Random Forest) consistently performed strongest, and explainability (e.g. SHAP) surfaced actionable, non-obvious predictors.
- **Ethics checklist before deployment**: Transparency (do students know?), Equity/Bias (does it perform equally across subgroups?), the "Care" Mandate (support vs. surveillance/punishment).
