# 8.1 Introduction to Machine Learning with Survey Data
- Survey data adds the "Human Element" administrative records miss
- Same supervised learning concepts, new feature source
- Literature: ensembles (Random Forest) consistently win

## Why Survey Data in ML Workflows
Administrative records tell you *what* happened (GPA, enrollment). Survey data helps explain *how it felt and why* — engagement, belonging, stress, perceived support. These constructs are often early indicators of outcomes, before they show up in GPA.

## Supervised Learning Recap, Survey Context
- **Target ($y$)**: academic success/failure, attitude categories, performance scores, persistence
- **Features ($X$)**: Likert responses (numeric), binary items, one-hot categoricals, composite scores, text vectors (Module 7)
- **Algorithm families**: Logistic Regression (interpretable), Decision Trees (intuitive), Random Forest/XGBoost (best predictive power), SVM, MLP (highest capacity, least interpretable)

## What the Literature Shows
Across real published studies (academic performance prediction, early-warning systems, student stress classification), **ensemble methods — especially Random Forest — consistently perform strongest**. Explainability tools (like SHAP) surface actionable, non-obvious predictors (e.g. perceived safety, sleep quality).

## Ethics Checklist Before Deployment
1. **Transparency** — do students know their feedback feeds predictive models?
2. **Equity/Bias** — does the model perform equally across demographic subgroups?
3. **The "Care" Mandate** — are predictions used for support, not surveillance or punishment?
